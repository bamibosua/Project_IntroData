## 4. Load an optional API key

In Colab's Secrets panel (key icon), add `S2_API_KEY` and enable notebook access. This cell imports it into the runtime environment without displaying the value. If the secret is absent or inaccessible, the crawler uses its existing environment settings or public access.

```python
import os
from google.colab import userdata

try:
    api_key = userdata.get("S2_API_KEY")
except (userdata.SecretNotFoundError, userdata.NotebookAccessError):
    api_key = None

if api_key:
    os.environ["S2_API_KEY"] = api_key
    print("API key loaded from Colab Secrets.")
else:
    print("No accessible Colab API key; using environment settings or public access.")
```
